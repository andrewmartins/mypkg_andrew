# 1 - If, elif, else

There are several conditional structures in Python. Firstly, if allows you to perform actions if a condition is met. In the example, we define that x=10, and we say if x is less than 3 then we want to display “It was true!” when we execute nothing is displayed because 10 is greater than 3.

````python
loc = 'Bank'

if loc == 'Auto Shop':
    print('Welcome to the Auto Shop!')
elif loc == 'Bank':
    print('Welcome to the bank!')
else:
    print('Where are you?')
````

# 7 - While

The while conditional structure executes instructions until the specified condition is no longer met. Here, as long as x is less than 10, we display the current value of 10, indicate that we're adding another 1 and perform the calculation.

````python
x = 0

while x < 10:
    print('x is currently: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
````

# 8 - Break, pass, continue

There are keywords for loops and while, continue allows you to continue the program without executing the remaining lines of the program, here for example the line 'x is still less than 10, adding 1 to x' is only executed when x is equal to 3 otherwise the program avoids the last lines each time.

break stops the loop.

Finally, pass is especially useful for developers: when you're developing, you sometimes know that you're going to have to develop certain functions, but you prefer to develop other things first and test them anyway, even if the program isn't entirely finished. So pass allows you to pass an instruction that is empty, because you can't just put the for statement without putting anything inside it - you'll get an error.

````python
x = 0

while x < 10:
    x+=1
    print('x is currently: ',x)
    if x==3:
        print('x==3')
    else:
        continue
    print(' x is still less than 10, adding 1 to x')
````

````python
x = 0

while x < 10:
    print('x is currently: ',x)
    print(' x is still less than 10, adding 1 to x')
    x+=1
    if x==3:
        print('Breaking because x==3')
        break
    else:
        continue
````
