# 1 - Opening a file : read mode

In the first session we saw how to create a small text file directly from a notebook with %%writefile. Let's create one to work with during this session.

````python
%%writefile hello.txt
Hello everyone !
This is the second line.
And this is the third one.
````

To read the contents of a file, we use the open() function. Its first argument is the path of the file, and the second is the mode: "r" for read mode. Read mode is the default, so open(path) and open(path,"r") are the same thing.

The variable returned by open() is a file object: it does not contain the text yet, it is a "handle" that lets us read from (or write to) the file.

````python
path="hello.txt"
````

````python
file=open(path,"r")
````

````python
file
````

## 1.1 - read

The read() method reads the entire file at once and returns it as a single string.

````python
content=file.read()
print(content)
````

If you call read() a second time, nothing is returned: the file object remembers where you stopped reading (the "cursor" is at the end of the file). Use seek(0) to move the cursor back to the beginning.

````python
file.read()
````

````python
file.seek(0)
file.read()
````

## 1.2 - readlines

The readlines() method reads the file line by line and returns a list, with one string per line. Note that each string still contains the line break "\n" at the end.

````python
file.seek(0)
file.readlines()
````

Since readlines() returns a list, you can iterate over it with a for loop.

````python
file.seek(0)
for line in file.readlines():
    print(line)
````

Each line is printed with an empty line after it: print() adds a line break, and the line itself already ends with "\n". The strip() method removes the spaces and line breaks at the beginning and end of a string.

````python
file.seek(0)
for line in file.readlines():
    print(line.strip())
````

## 1.3 - Closing a file

Always close a file when you have finished manipulating it with the close() method. A file that is not closed can lock the file for other programs, and changes written to it may not be saved.

````python
file.close()
````

Once a file is closed, you can no longer read from it.

````python
file.read()
````

# 2 - Opening a file : write mode

To write to a file, use the write mode "w". The write() method writes a string inside the file. Just as with print(), you must specify a string, or use str(variable) to transform your variable into one.

**Warning:** write mode deletes everything the file contained before. If the file does not exist, it is created.

````python
path="new_file.txt"
````

````python
file=open(path,"w") # "w" mode removes everything from the file if it already existed
````

````python
file.write("Hello everyone !")
````

The number displayed is the number of characters that were written. Notice that write() does not add a line break for you: if you want one, you must add "\n" yourself.

````python
file.write("\nThis is a second line.")
````

````python
file.close()
````

Let's check what the file contains by opening it again in read mode.

````python
file=open(path,"r")
print(file.read())
file.close()
````

If we open the same file in write mode again and write something else, the previous contents are lost.

````python
file=open(path,"w")
file.write("Everything else is gone.")
file.close()
````

````python
file=open(path,"r")
print(file.read())
file.close()
````

# 3 - Opening a file : append mode

Unlike write mode, the append mode "a" does not delete the information contained in the file. New content is added at the end of the existing content. This is the mode to use when you want to add lines to a file, for example a log or a list of results.

````python
path="new_file.txt"
file=open(path,"a")
file.write("\nI'm adding just a little sentence.")
file.close()
````

````python
file=open(path,"r")
print(file.read())
file.close()
````

Here is a summary of the three modes:

| Mode | Name | Existing content | File does not exist |
|---|---|---|---|
| "r" | read | kept, read only | error |
| "w" | write | **deleted** | created |
| "a" | append | kept, new content added at the end | created |

# 4 - The with statement

Forgetting to close a file is a very common mistake. Python provides the with statement to avoid it: the file is opened at the beginning of the block and automatically closed when the block ends, even if an error occurs inside it.

The syntax is: with open(path, mode) as file, followed by an indented block containing the instructions that use the file.

````python
with open("hello.txt","r") as file:
    print(file.read())
````

Once we leave the block, the file is closed.

````python
file.closed
````

The with statement works with all modes. From now on, this is the way you should open files.

````python
with open("new_file.txt","w") as file:
    file.write("Written with the with statement.")
````

````python
with open("new_file.txt","a") as file:
    file.write("\nAppended with the with statement.")
````

````python
with open("new_file.txt","r") as file:
    for line in file:
        print(line.strip())
````

Notice that in the last example we iterate directly over the file object with for line in file, without calling readlines(). This is the most common way of going through a file line by line, and it works even for very large files because lines are read one at a time instead of loading the whole file in memory.

# 6 - Structured text files

Data is very often stored in text files where each line is a record and each value is separated by a delimiter, usually a comma: this is the CSV format (Comma-Separated Values). Let's create a small one.

````python
%%writefile patients.csv
name,age,city
Alice,34,Bordeaux
Bruno,52,Paris
Chloe,27,Lyon
````

Each line is a string. To retrieve the individual values, we use the split() method seen in the first session, giving it the delimiter as argument. The result is a list of strings.

````python
with open("patients.csv","r") as file:
    for line in file:
        print(line.strip().split(","))
````

The first line is the header: it contains the column names, not data. A common pattern is to read the header separately with readline(), which reads a single line, and then loop over the remaining lines.

````python
with open("patients.csv","r") as file:
    header=file.readline().strip().split(",")
    for line in file:
        values=line.strip().split(",")
        print(header, values)
````

Remember that everything read from a file is a string, exactly as with input(). If you want to use the ages as numbers, you must convert them with int() or float().

````python
total_age=0
number_of_patients=0

with open("patients.csv","r") as file:
    file.readline() # skip the header
    for line in file:
        values=line.strip().split(",")
        total_age+=int(values[1])
        number_of_patients+=1

print("Mean age : %.1f"%(total_age/number_of_patients))
````

Combining files, loops, dictionaries and conditions, you can now summarise a data file. Here we count how many patients live in each city.

````python
cities={}

with open("patients.csv","r") as file:
    file.readline()
    for line in file:
        values=line.strip().split(",")
        city=values[2]
        if city not in cities:
            cities[city]=0
        cities[city]+=1

print(cities)
````
